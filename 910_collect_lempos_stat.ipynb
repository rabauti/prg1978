```python
from datetime import datetime

from data_helpers.syntax_graph import SyntaxGraph
from data_helpers.tasak_reader import TasakReader
from data_helpers.utils import ListUtils
```

```python
%%time

file_name = 'data/tasak.vert'

my_reader = TasakReader(
   file_name = file_name
)

stats = {}

date_time = datetime.now().strftime("%Y%m%d-%H%M%S")


# hoitakse algusest lõpuni mälus
lemmas_stat = {}
lemmas_total = 0


count = 0
for collection_id, graph in my_reader.get_sentences():
    # matrix for node distances
    #dpath = graph.get_distances_matrix()
    
    for node in graph.nodes():
        if not node: 
            continue
        key = '%s\t%s' % (graph.nodes[node]['lemma'], graph.nodes[node]['POS'],)
        if not key in lemmas_stat:
            lemmas_stat[key] = 1
        else:
            lemmas_stat[key] += 1
        
        lemmas_total += 1
        
print('Lemmas total:', lemmas_total)
```

```python
import pandas as pd
```

```python
data = [(key.split('\t')[0], key.split('\t')[1], value,) for key, value in lemmas_stat.items() ]
df_plain = pd.DataFrame(data, columns=["lemma", "POS", "total"])
df_plain.sort_values(["total", "lemma", "POS"], ascending=[False, True, True], inplace=True)
df_plain.head()

df_plain.to_csv(f"tasak_lemmas_{date_time}.tsv", index=None, sep="\t")
```
