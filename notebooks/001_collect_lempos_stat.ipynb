# Õpikulausete korpuse lemma + pos statistika kogumine

Kogutakse kokku statistika:

- lemmade esinemised <code>opik_lemmas_YYYYMMDD.tsv</code>;
- lemmade koguarv.


```python
from notebook_context import corpus_reader, LISTS_FOLDER
from src.syntax.syntax_graph import SyntaxGraph
from src.syntax.utils import ListUtils
import pandas as pd

LEMMAS_LIST = LISTS_FOLDER / "stats/lemmas.tsv"
```

```python
%%time


stats = {}
# hoitakse algusest lõpuni mälus
lemmas_stat = {}
lemmas_total = 0


count = 0
for collection_id, graph in corpus_reader.get_sentences():
  # matrix for node distances
  #dpath = graph.get_distances_matrix()
  
  for node in graph.nodes():
    if not node:
        continue
    key = '%s\t%s' % (graph.nodes[node]['lemma'], graph.nodes[node]['POS'],)
    if key not in lemmas_stat:
     lemmas_stat[key] = 1
    else:
        lemmas_stat[key] += 1
    
    lemmas_total += 1
    
print('Lemmas total:', lemmas_total)
```

```python
data = [(key.split('\t')[0], key.split('\t')[1], value,) for key, value in lemmas_stat.items() ]
df_plain = pd.DataFrame(data, columns=["lemma", "POS", "total"])
df_plain.sort_values(["total", "lemma", "POS"], ascending=[False, True, True], inplace=True)
df_plain.head()

df_plain.to_csv(LEMMAS_LIST, index=None, sep="\t")
```
