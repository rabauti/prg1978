```python
from notebook_context import corpus_reader

levels_tokens = {}
tokens_total = 0
sentences_total = 0

print("begin")
for sent_id, graph in corpus_reader.get_sentences():
    sentences_total += 1
    tokens_total += graph.number_of_nodes()
    level = graph.get_metadata("sent_level")
    if level not in levels_tokens:
        levels_tokens[level] = 0
    levels_tokens[level] += graph.number_of_nodes()
print("done.")


display(f"sentences_total: {sentences_total}")
display(f"tokens_total: {tokens_total}")
display(levels_tokens)
```

```python
ids = []
print("begin")
sentences_total = 0
for sent_id, graph in corpus_reader.get_sentences():
    tokens_total += graph.number_of_nodes()
    row_nr = graph.get_metadata("row_nr")
    if row_nr in ids:
        print(row_nr, sentences_total, graph.get_metadata())
    else:
        ids.append(row_nr)
    sentences_total += 1
print("done.")
```

```python
import pandas as pd
df_wordstat = pd.DataFrame.from_dict([{'level': level, 'count': levels_tokens[level]}for level in levels_tokens.keys()] )
df_wordstat.head(100)
df_wordstat.to_csv('corp_tokens_stat.csv', index=None)
```
